```python
import ast

def convert(text):
    L = []

    # convert string to list of dictionaries
    if isinstance(text, str):
        text = ast.literal_eval(text)

    for i in text:
        # handle both dicts and strings
        if isinstance(i, dict):
            L.append(i['name'])
        else:
            L.append(i)

    return L

# Apply the corrected function
movies['genres'] = movies['genres'].apply(convert)  # applying corrected function
```

```python
movies.dropna(inplace=True)
```

```python

def convert3(text):
    L = []
    counter = 0

    # convert string to list if needed
    if isinstance(text, str):
        text = ast.literal_eval(text)

    for i in text:
        if counter < 3:
            if isinstance(i, dict):
                L.append(i['name'])
            else:
                L.append(i)
        counter += 1

    return L

# Apply the function
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
movies['cast'][0]
```

```python
def fetch_director(text):
    L = []

    # convert string to list if needed
    if isinstance(text, str):
        text = ast.literal_eval(text)

    for i in text:
        if isinstance(i, dict):
            if i.get('job') == 'Director':
                L.append(i.get('name'))

    return L
    
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

# vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer 
```

```python
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()
```

```python
#we will apply steming to covert the similiar words into one
# for ex[love,loving,love]
# output: [love,love,love]
```

```python
def stem(text):#to  convert similiar words into single one using stemming
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

# main function 

```python
def recommend(movie):
    movie_index=new_df[new_df['title']==movie].index[0]#index of each movie in dataframe
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]#lambda function shows that we have to do sortingbase on second number in each array of list which is a distance and with the help of [1:6] we extract top 5 movies havig closest distance
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
    
    
```
